```python
##Bike Management
# #Functions created to add new bikes

def add_new_bikes(bike_fleet):      
    print("click to add new bike to bike_fleet")

    if len(bike_fleet)>0:   #len() counts items that asks the question "Is the list empty or does it have at least one bike in it?"
        new_bike_id=max(bike["Bike_ID"] for bike in bike_fleet)+1
    else: 
        new_bike_id=101
    print(f"New Bike ID created:{new_bike_id}")

    Bike_Type=input("Insert Bike Type (Mountain, Road, Hybrid): ")

    Bike_Availability= "Yes" #default availability for all new bikes is Yes

    new_bikes={"Bike_ID": new_bike_id, "Bike_Type":Bike_Type, "Bike_Availability": Bike_Availability}

    bike_fleet.append(new_bikes)

    print(f"Successfully Added New Bike {new_bike_id} of {Bike_Type}")
```
